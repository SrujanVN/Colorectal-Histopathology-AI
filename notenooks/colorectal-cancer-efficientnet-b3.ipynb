```python
# =============================================================================
# Colorectal Histopathology Classification (EfficientNet-B3) | Kather CRC-2016
# -----------------------------------------------------------------------------
# Pipeline overview:
#   1. Imports
#   2. Configuration, paths, colour palette
#   3. Reproducibility, device and path resolution
#   4. Dataset and per-split transforms
#   5. Data loaders (70 / 15 / 15 split)
#   6. Model (EfficientNet-B3 backbone + custom classification head)
#   7. Training utilities (early stopping, epoch runner, training loop)
#   8. Evaluation
#   9. Plotting (history, confusion matrix, ROC)
#  10. Main entry point
# =============================================================================

# ----------------------------- 1. IMPORTS ------------------------------------
import os
import gc
import random
import warnings
from collections import Counter

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset
from torch.optim.lr_scheduler import CosineAnnealingLR
from torchvision import transforms, models

from sklearn.metrics import (classification_report, confusion_matrix,
                             cohen_kappa_score, roc_curve, auc)
from sklearn.preprocessing import label_binarize

warnings.filterwarnings("ignore")


# ------------------------- 2. CONFIGURATION ----------------------------------
SEED = 42

# Data split
TRAIN_RATIO = 0.70
VAL_RATIO = 0.15               # test gets the remainder (0.15)

# Training hyperparameters
BATCH_SIZE = 32
NUM_EPOCHS = 40
IMAGE_SIZE = 224
LEARNING_RATE = 0.001
WEIGHT_DECAY = 0.01
DROPOUT_RATE = 0.5

# Early stopping
PATIENCE = 10
MIN_DELTA = 0.001

# Cap on UNKNOWN images so they don't dominate the dataset
MAX_UNKNOWN = 500

# Output files
BEST_MODEL_PATH = "best_colorectal_efficientnet_b3.pth"
FINAL_MODEL_PATH = "final_colorectal_efficientnet_b3.pth"

# Dataset locations (exact Kaggle paths; auto-detection is used as a fallback)
KAGGLE_INPUT = "/kaggle/input"
DATASET_DIR = ("/kaggle/input/datasets/kmader/colorectal-histology-mnist/"
               "kather_texture_2016_image_tiles_5000/"
               "Kather_texture_2016_image_tiles_5000")
UNSPLASH_DIR = ("/kaggle/input/datasets/lprdosmil/unsplash-random-images-collection/"
                "unsplash-images-collection")

# Class names (UNKNOWN is appended only when Unsplash images are available)
TISSUE_CLASSES = ["01_TUMOR", "02_STROMA", "03_COMPLEX", "04_LYMPHO",
                  "05_DEBRIS", "06_MUCOSA", "07_ADIPOSE", "08_EMPTY"]
UNKNOWN_CLASS = "UNKNOWN"
IMG_EXTS = (".png", ".jpg", ".jpeg", ".tiff", ".tif")

# Colour palette shared by every plot (edit here to restyle all graphs)
PALETTE = {
    "train": "#0F9D8A",     # teal   -> training curves
    "val": "#F26B3A",       # orange -> validation curves
    "chance": "#5C5C5C",    # grey   -> random-classifier baseline
}
ROC_COLORS = ["#E63946", "#F4A261", "#E9C46A", "#2A9D8F", "#264653",
              "#8E44AD", "#1D8FE1", "#6A994E", "#7F7F7F"]
CM_CMAP = "YlGnBu"          # confusion-matrix colour map


# ------------------- 3. REPRODUCIBILITY, DEVICE, PATHS -----------------------
def set_seed(seed=SEED):
    """Fix all random seeds so runs are repeatable."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False


set_seed()

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")


def find_tissue_root(configured_path, base=KAGGLE_INPUT):
    """
    Return the folder that directly contains the class sub-folders
    (01_TUMOR, 02_STROMA, ...). Uses the configured path if valid,
    otherwise searches everything under /kaggle/input.
    """
    if os.path.isdir(os.path.join(configured_path, "01_TUMOR")):
        return configured_path
    for root, dirs, _ in os.walk(base):
        if "01_TUMOR" in dirs:
            print(f"Auto-detected tissue dataset at: {root}")
            return root
    raise FileNotFoundError(
        f"Could not find '01_TUMOR' under {base}. Check that the dataset is attached.")


def find_unknown_root(configured_path, base=KAGGLE_INPUT):
    """
    Return a folder of random images for the UNKNOWN class, or None if absent.
    Falls back to searching for a folder with 'unsplash' in its path that
    holds images directly.
    """
    if configured_path and os.path.isdir(configured_path):
        return configured_path
    for root, _, files in os.walk(base):
        if "unsplash" in root.lower() and any(f.lower().endswith(IMG_EXTS) for f in files):
            print(f"Auto-detected UNKNOWN images at: {root}")
            return root
    print("WARNING: no Unsplash folder found - training with the 8 Kather classes only.")
    return None


# ---------------------------- 4. DATASET -------------------------------------
class ColorectalCancerDataset(Dataset):
    """
    Index of (image_path, label) pairs for the 8 tissue classes plus an
    optional UNKNOWN class built from a flat folder of random photos.
    Images are loaded lazily by the split wrapper (TransformSubset).
    """

    def __init__(self, root_dir, unknown_dir=None, max_unknown=None):
        self.root_dir = root_dir
        self.samples = []                    # list of (image_path, label)
        self.class_counts = Counter()

        # Build the class list first so labels are always contiguous (0..N-1)
        has_unknown = unknown_dir is not None and os.path.isdir(unknown_dir)
        self.classes = list(TISSUE_CLASSES) + ([UNKNOWN_CLASS] if has_unknown else [])
        self.class_to_idx = {c: i for i, c in enumerate(self.classes)}

        self._load_tissue_classes()
        if has_unknown:
            self._load_unknown_class(unknown_dir, max_unknown)

        if not self.samples:
            raise RuntimeError(f"No images found under '{root_dir}'. Check the dataset path.")
        self._print_summary()

    @staticmethod
    def _list_images(folder):
        """Sorted list of image files in a folder (sorted for reproducibility)."""
        return [os.path.join(folder, f) for f in sorted(os.listdir(folder))
                if f.lower().endswith(IMG_EXTS)]

    def _load_tissue_classes(self):
        """Read images from each of the 8 class sub-folders."""
        for class_name in TISSUE_CLASSES:
            class_path = os.path.join(self.root_dir, class_name)
            if not os.path.exists(class_path):
                print(f"WARNING: Missing class directory: {class_path}")
                continue
            items = [(p, self.class_to_idx[class_name]) for p in self._list_images(class_path)]
            self.samples.extend(items)
            self.class_counts[class_name] = len(items)

    def _load_unknown_class(self, unknown_dir, max_unknown):
        """Read the flat folder of random images and label them UNKNOWN."""
        paths = self._list_images(unknown_dir)
        if max_unknown is not None:
            paths = paths[:max_unknown]
        items = [(p, self.class_to_idx[UNKNOWN_CLASS]) for p in paths]
        self.samples.extend(items)
        self.class_counts[UNKNOWN_CLASS] = len(items)

    def _print_summary(self):
        print("\nDataset initialized.")
        print("-" * 70)
        for name in self.classes:
            print(f"  {name:<15}: {self.class_counts.get(name, 0)}")
        print("-" * 70)
        print(f"Total samples : {len(self.samples)}")
        print(f"Active classes: {self.classes}\n")

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        return self.samples[idx]


class TransformSubset(Dataset):
    """
    A split (train/val/test) that owns its OWN transform.
    Avoids the shared-transform bug of random_split, where changing
    `dataset.transform` for one split silently changes it for all splits.
    """

    def __init__(self, dataset, indices, transform=None):
        self.dataset = dataset
        self.indices = indices
        self.transform = transform

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, idx):
        img_path, label = self.dataset.samples[self.indices[idx]]
        try:
            image = Image.open(img_path).convert("RGB")
        except Exception as e:
            raise RuntimeError(f"Failed to load image: {img_path}\nError: {e}")
        if self.transform:
            image = self.transform(image)
        return image, label


def get_transforms():
    """Augmented transform for training; plain resize + normalise for eval."""
    normalize = transforms.Normalize(mean=[0.485, 0.456, 0.406],
                                     std=[0.229, 0.224, 0.225])   # ImageNet stats

    train_transform = transforms.Compose([
        transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomVerticalFlip(p=0.5),
        transforms.RandomRotation(degrees=15),
        transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.1),
        transforms.RandomAffine(degrees=0, translate=(0.1, 0.1), scale=(0.9, 1.1)),
        transforms.RandomApply([transforms.GaussianBlur(kernel_size=3)], p=0.2),
        transforms.ToTensor(),
        normalize,
    ])
    eval_transform = transforms.Compose([
        transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
        transforms.ToTensor(),
        normalize,
    ])
    return train_transform, eval_transform


# ------------------------- 5. DATA LOADERS -----------------------------------
def create_data_loaders(dataset_dir, batch_size=32, train_ratio=0.70, val_ratio=0.15,
                        unknown_dir=None, max_unknown=None):
    """Build train / validation / test loaders using a reproducible 70-15-15 split."""
    print("=" * 70)
    print("CREATING DATA LOADERS")
    print("=" * 70)

    train_transform, eval_transform = get_transforms()
    full_dataset = ColorectalCancerDataset(dataset_dir, unknown_dir, max_unknown)

    # Split sizes (test takes the remainder so no sample is lost to rounding)
    total = len(full_dataset)
    train_size = int(train_ratio * total)
    val_size = int(val_ratio * total)
    test_size = total - train_size - val_size
    print(f"Train: {train_size} | Val: {val_size} | Test: {test_size}")

    # Shuffle indices once with a fixed seed, then slice into the three splits
    generator = torch.Generator().manual_seed(SEED)
    indices = torch.randperm(total, generator=generator).tolist()
    train_idx = indices[:train_size]
    val_idx = indices[train_size:train_size + val_size]
    test_idx = indices[train_size + val_size:]

    # Each split gets its own transform (augmentation only for training)
    train_ds = TransformSubset(full_dataset, train_idx, train_transform)
    val_ds = TransformSubset(full_dataset, val_idx, eval_transform)
    test_ds = TransformSubset(full_dataset, test_idx, eval_transform)

    pin = torch.cuda.is_available()
    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True,
                              num_workers=2, pin_memory=pin, drop_last=True,
                              persistent_workers=True)
    val_loader = DataLoader(val_ds, batch_size=batch_size, shuffle=False,
                            num_workers=2, pin_memory=pin,
                            persistent_workers=True)
    test_loader = DataLoader(test_ds, batch_size=batch_size, shuffle=False,
                             num_workers=2, pin_memory=pin,
                             persistent_workers=True)

    print("Data loaders created successfully.\n")
    return train_loader, val_loader, test_loader, full_dataset.classes


# ------------------------------ 6. MODEL -------------------------------------
class EfficientNetB3Classifier(nn.Module):
    """EfficientNet-B3 (ImageNet pretrained) with a custom multi-layer head."""

    def __init__(self, num_classes=9, dropout_rate=0.5):
        super().__init__()
        print("Loading EfficientNet-B3...")

        # Try pretrained weights; fall back to random init if download fails
        try:
            self.backbone = models.efficientnet_b3(weights="IMAGENET1K_V1")
            print("ImageNet pretrained weights loaded.")
        except Exception as e:
            print(f"Could not load pretrained weights: {e}")
            print("Loading model without pretrained weights.")
            self.backbone = models.efficientnet_b3(weights=None)

        # Freeze everything except the last 30 parameter tensors (stable fine-tuning)
        for param in list(self.backbone.parameters())[:-30]:
            param.requires_grad = False

        # Replace the original classifier with a deeper, regularised head
        in_features = self.backbone.classifier[1].in_features
        self.backbone.classifier = nn.Sequential(
            nn.Dropout(p=dropout_rate),
            nn.Linear(in_features, 512),
            nn.ReLU(inplace=True),
            nn.BatchNorm1d(512),
            nn.Dropout(p=dropout_rate * 0.5),
            nn.Linear(512, 256),
            nn.ReLU(inplace=True),
            nn.BatchNorm1d(256),
            nn.Dropout(p=dropout_rate * 0.3),
            nn.Linear(256, num_classes),
        )

    def forward(self, x):
        return self.backbone(x)


# ------------------------ 7. TRAINING UTILITIES ------------------------------
class EarlyStopping:
    """Stop training when validation loss stops improving."""

    def __init__(self, patience=10, min_delta=0.001, restore_best_weights=True):
        self.patience = patience
        self.min_delta = min_delta
        self.restore_best_weights = restore_best_weights
        self.best_loss = None
        self.counter = 0
        self.best_weights = None

    def __call__(self, val_loss, model):
        if self.best_loss is None or val_loss < self.best_loss - self.min_delta:
            # Improvement -> remember this state and reset the patience counter
            self.best_loss = val_loss
            self.counter = 0
            self.save_checkpoint(model)
        else:
            self.counter += 1

        if self.counter >= self.patience:
            if self.restore_best_weights and self.best_weights is not None:
                model.load_state_dict(self.best_weights)
            return True
        return False

    def save_checkpoint(self, model):
        self.best_weights = {k: v.detach().clone() for k, v in model.state_dict().items()}


def run_epoch(model, loader, criterion, device, desc, optimizer=None):
    """
    One pass over `loader`.
    With an optimizer -> training mode; without -> evaluation mode.
    Returns (average_loss, accuracy_percent).
    """
    gc.collect()  
    training = optimizer is not None
    model.train() if training else model.eval()

    total_loss, correct, total = 0.0, 0, 0
    pbar = tqdm(loader, desc=desc)

    with torch.set_grad_enabled(training):
        for data, target in pbar:
            data = data.to(device, non_blocking=True)
            target = target.to(device, non_blocking=True)

            if training:
                optimizer.zero_grad()

            output = model(data)
            loss = criterion(output, target)

            if training:
                loss.backward()
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)  # avoid exploding gradients
                optimizer.step()

            total_loss += loss.item()
            correct += (output.argmax(dim=1) == target).sum().item()
            total += target.size(0)
            pbar.set_postfix({"Loss": f"{loss.item():.4f}",
                              "Acc": f"{100.0 * correct / total:.2f}%"})

    return total_loss / max(len(loader), 1), 100.0 * correct / max(total, 1)


def train_model(model, train_loader, val_loader, device, num_epochs=40):
    """Full training loop: label smoothing, AdamW, cosine LR and early stopping."""
    criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
    optimizer = optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
    scheduler = CosineAnnealingLR(optimizer, T_max=num_epochs, eta_min=1e-6)
    early_stopping = EarlyStopping(patience=PATIENCE, min_delta=MIN_DELTA)

    history = {"train_loss": [], "train_acc": [], "val_loss": [], "val_acc": []}
    best_val_acc = 0.0

    print("\n" + "=" * 70)
    print("STARTING TRAINING")
    print("=" * 70)

    for epoch in range(num_epochs):
        tag = f"Epoch {epoch + 1}/{num_epochs}"

        # --- training pass, then validation pass ---
        train_loss, train_acc = run_epoch(model, train_loader, criterion, device,
                                          f"{tag} [Train]", optimizer)
        val_loss, val_acc = run_epoch(model, val_loader, criterion, device,
                                      f"{tag} [Val]")

        # --- record metrics ---
        history["train_loss"].append(train_loss)
        history["train_acc"].append(train_acc)
        history["val_loss"].append(val_loss)
        history["val_acc"].append(val_acc)

        print(f"\n{tag} results:")
        print(f"   Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}%")
        print(f"   Val   Loss: {val_loss:.4f} | Val   Acc: {val_acc:.2f}%")

        # --- checkpoint the best model by validation accuracy ---
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            torch.save(model.state_dict(), BEST_MODEL_PATH)
            print(f"   New best model saved: {best_val_acc:.2f}%")

        scheduler.step()

        # --- stop early if validation loss has plateaued ---
        if early_stopping(val_loss, model):
            print(f"Early stopping at epoch {epoch + 1}")
            break

        # --- periodic memory cleanup ---
        if epoch % 10 == 0:
            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

        print("-" * 70)

    return history, best_val_acc


# --------------------------- 8. EVALUATION -----------------------------------
def evaluate_model(model, test_loader, device):
    """Return predictions, ground-truth labels and class probabilities."""
    model.eval()
    all_preds, all_targets, all_probs = [], [], []

    with torch.no_grad():
        for data, target in tqdm(test_loader, desc="Evaluating"):
            data = data.to(device, non_blocking=True)
            output = model(data)
            probs = torch.softmax(output, dim=1)

            all_preds.extend(output.argmax(dim=1).cpu().numpy())
            all_targets.extend(target.numpy())
            all_probs.extend(probs.cpu().numpy())

    return np.array(all_preds), np.array(all_targets), np.array(all_probs)


# ----------------------------- 9. PLOTTING -----------------------------------
def plot_training_history(history):
    """Side-by-side accuracy and loss curves for train vs validation."""
    plt.style.use("seaborn-v0_8-whitegrid")
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    epochs = range(1, len(history["train_acc"]) + 1)

    # Accuracy panel
    ax1.plot(epochs, history["train_acc"], color=PALETTE["train"], lw=2.5,
             marker="o", ms=4, label="Training Accuracy")
    ax1.plot(epochs, history["val_acc"], color=PALETTE["val"], lw=2.5,
             marker="s", ms=4, label="Validation Accuracy")
    ax1.set_title("Training and Validation Accuracy", fontsize=14, fontweight="bold")
    ax1.set_xlabel("Epoch", fontsize=12)
    ax1.set_ylabel("Accuracy (%)", fontsize=12)
    ax1.legend(fontsize=10)

    # Loss panel
    ax2.plot(epochs, history["train_loss"], color=PALETTE["train"], lw=2.5,
             marker="o", ms=4, label="Training Loss")
    ax2.plot(epochs, history["val_loss"], color=PALETTE["val"], lw=2.5,
             marker="s", ms=4, label="Validation Loss")
    ax2.set_title("Training and Validation Loss", fontsize=14, fontweight="bold")
    ax2.set_xlabel("Epoch", fontsize=12)
    ax2.set_ylabel("Loss", fontsize=12)
    ax2.legend(fontsize=10)

    plt.tight_layout()
    plt.show()


def plot_confusion_matrix(y_true, y_pred, class_names):
    """Heatmap of true vs predicted labels (all classes shown, even if absent)."""
    cm = confusion_matrix(y_true, y_pred, labels=list(range(len(class_names))))

    plt.figure(figsize=(12, 10))
    sns.heatmap(cm, annot=True, fmt="d", cmap=CM_CMAP,
                linewidths=0.5, linecolor="white",
                xticklabels=class_names, yticklabels=class_names,
                cbar_kws={"label": "Number of Samples"})
    plt.title("Confusion Matrix", fontsize=16, fontweight="bold", pad=20)
    plt.xlabel("Predicted Label", fontsize=12)
    plt.ylabel("True Label", fontsize=12)
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
    return cm


def plot_roc_curves(y_true, y_probs, class_names):
    """One-vs-rest ROC curve per class; returns the mean AUC over valid classes."""
    n_classes = len(class_names)
    y_true_bin = label_binarize(y_true, classes=list(range(n_classes)))

    plt.figure(figsize=(14, 10))
    mean_auc, valid_classes = 0.0, 0

    for i, class_name in enumerate(class_names):
        positives = y_true_bin[:, i].sum()

        # ROC is undefined if a class has no positives or no negatives
        if positives == 0 or positives == len(y_true_bin):
            print(f"Skipping ROC for {class_name}: needs both positive and negative samples.")
            continue

        fpr, tpr, _ = roc_curve(y_true_bin[:, i], y_probs[:, i])
        class_auc = auc(fpr, tpr)
        mean_auc += class_auc
        valid_classes += 1

        plt.plot(fpr, tpr, color=ROC_COLORS[i % len(ROC_COLORS)], lw=2.5,
                 label=f"{class_name} (AUC = {class_auc:.3f})")

    mean_auc = mean_auc / valid_classes if valid_classes else float("nan")

    # Diagonal = performance of random guessing
    plt.plot([0, 1], [0, 1], linestyle="--", lw=2, color=PALETTE["chance"],
             label="Random Classifier (AUC = 0.500)")
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel("False Positive Rate", fontsize=12)
    plt.ylabel("True Positive Rate", fontsize=12)
    plt.title(f"Multi-class ROC Curves\nMean AUC = {mean_auc:.3f}",
              fontsize=14, fontweight="bold")
    plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left", fontsize=10)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
    return mean_auc


# ------------------------------ 10. MAIN -------------------------------------
def describe_kappa(kappa):
    """Translate Cohen's Kappa into a plain-English agreement level."""
    if kappa > 0.8:
        return "Excellent agreement"
    if kappa > 0.6:
        return "Substantial agreement"
    if kappa > 0.4:
        return "Moderate agreement"
    return "Poor agreement"


def main():
    print("\n" + "=" * 70)
    print("COLORECTAL HISTOPATHOLOGY CLASSIFICATION | EfficientNet-B3")
    print("=" * 70)
    print(f"Using device: {device}")
    if torch.cuda.is_available():
        print(f"GPU: {torch.cuda.get_device_name(0)}")

    # ---- Resolve dataset locations ----
    tissue_dir = find_tissue_root(DATASET_DIR)
    unknown_dir = find_unknown_root(UNSPLASH_DIR)
    print(f"Tissue dataset : {tissue_dir}")
    print(f"UNKNOWN dataset: {unknown_dir}")

    # ---- Data ----
    train_loader, val_loader, test_loader, class_names = create_data_loaders(
        tissue_dir, BATCH_SIZE, TRAIN_RATIO, VAL_RATIO,
        unknown_dir=unknown_dir,
        max_unknown=MAX_UNKNOWN if unknown_dir else None,
    )

    # ---- Model ----
    print("=" * 70)
    print("BUILDING MODEL")
    print("=" * 70)
    model = EfficientNetB3Classifier(num_classes=len(class_names),
                                     dropout_rate=DROPOUT_RATE).to(device)
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"Total parameters    : {total_params:,}")
    print(f"Trainable parameters: {trainable_params:,}")
    print(f"Classes ({len(class_names)}): {class_names}")

    # ---- Training ----
    history, best_val_acc = train_model(model, train_loader, val_loader, device, NUM_EPOCHS)

    # Reload the best checkpoint (by validation accuracy) before testing
    if os.path.exists(BEST_MODEL_PATH):
        print("\nLoading best model...")
        model.load_state_dict(torch.load(BEST_MODEL_PATH, map_location=device))
    print(f"Best validation accuracy: {best_val_acc:.2f}%")

    print("\nGenerating training curves...")
    plot_training_history(history)

    # ---- Test-set evaluation ----
    print("\n" + "=" * 70)
    print("TEST EVALUATION")
    print("=" * 70)
    y_pred, y_true, y_probs = evaluate_model(model, test_loader, device)

    test_accuracy = (y_pred == y_true).mean() * 100
    print(f"\nTest Accuracy: {test_accuracy:.2f}%")

    print("\nClassification Report")
    print("=" * 70)
    print(classification_report(y_true, y_pred,
                                labels=list(range(len(class_names))),
                                target_names=class_names,
                                digits=4, zero_division=0))

    kappa = cohen_kappa_score(y_true, y_pred)
    print(f"Cohen's Kappa: {kappa:.4f} -> {describe_kappa(kappa)}")

    # ---- Visualisations ----
    print("\nGenerating confusion matrix...")
    plot_confusion_matrix(y_true, y_pred, class_names)

    print("\nGenerating ROC curves...")
    mean_auc = plot_roc_curves(y_true, y_probs, class_names)

    # ---- Summary ----
    print("\n" + "=" * 70)
    print("FINAL RESULTS")
    print("=" * 70)
    print(f"Test Accuracy     : {test_accuracy:.2f}%")
    print(f"Mean AUC          : {mean_auc:.4f}")
    print(f"Cohen's Kappa     : {kappa:.4f}")
    print(f"Best Val Accuracy : {best_val_acc:.2f}%")
    print(f"Classes           : {len(class_names)}")

    # ---- Save final model together with its metadata ----
    torch.save({
        "model_name": "EfficientNet-B3",
        "model_state_dict": model.state_dict(),
        "class_names": class_names,
        "test_accuracy": test_accuracy,
        "mean_auc": mean_auc,
        "cohen_kappa": kappa,
        "best_val_accuracy": best_val_acc,
        "history": history,
        "seed": SEED,
        "image_size": IMAGE_SIZE,
        "batch_size": BATCH_SIZE,
        "learning_rate": LEARNING_RATE,
        "weight_decay": WEIGHT_DECAY,
        "num_epochs": NUM_EPOCHS,
    }, FINAL_MODEL_PATH)

    print("\nModels saved successfully:")
    print(f"  1. {BEST_MODEL_PATH}")
    print(f"  2. {FINAL_MODEL_PATH}")
    print("\n" + "=" * 70)
    print("TRAINING COMPLETE")
    print("=" * 70)


if __name__ == "__main__":
    main()
```
