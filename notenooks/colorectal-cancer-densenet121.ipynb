```python
# =============================================================================
# Colorectal Histology Classification (DenseNet121) with an UNKNOWN class
# -----------------------------------------------------------------------------
# Pipeline overview:
#   1. Configuration & reproducibility
#   2. Dataset (8 tissue classes + UNKNOWN from random Unsplash photos)
#   3. Transforms & data loaders (70 / 15 / 15 split)
#   4. Model (DenseNet121 backbone + custom classification head)
#   5. Training utilities (early stopping, training loop)
#   6. Evaluation & plotting
#   7. Main entry point
# =============================================================================

# ----------------------------- 1. IMPORTS ------------------------------------
import os
import gc
import warnings
from collections import Counter

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset, random_split
from torch.optim.lr_scheduler import CosineAnnealingLR
from torchvision import transforms, models

from sklearn.metrics import (classification_report, confusion_matrix,
                             cohen_kappa_score, roc_curve, auc)
from sklearn.preprocessing import label_binarize

warnings.filterwarnings('ignore')


# ------------------------- 2. CONFIG & REPRODUCIBILITY -----------------------
SEED = 42

# Dataset locations (Kaggle paths)
KAGGLE_INPUT = '/kaggle/input'

# Exact path taken from your file listing (note the lowercase outer folder name)
DATASET_DIR = ('/kaggle/input/datasets/kmader/colorectal-histology-mnist/'
               'kather_texture_2016_image_tiles_5000/'
               'Kather_texture_2016_image_tiles_5000')

# Exact Unsplash folder taken from your file listing
UNSPLASH_DIR = ('/kaggle/input/datasets/lprdosmil/unsplash-random-images-collection/'
                'unsplash-images-collection')

# Training hyperparameters
BATCH_SIZE = 32
NUM_EPOCHS = 40
MAX_UNKNOWN = 500          # cap on UNKNOWN images so they don't dominate

# Colour palette used across every plot (change here to restyle all graphs)
PALETTE = {
    'train': '#0F9D8A',    # teal   -> training curves
    'val':   '#F26B3A',    # orange -> validation curves
    'chance': '#5C5C5C',   # grey   -> random-classifier baseline
}
ROC_COLORS = ['#E63946', '#F4A261', '#E9C46A', '#2A9D8F', '#264653',
              '#8E44AD', '#1D8FE1', '#6A994E', '#7F7F7F']
CM_CMAP = 'YlGnBu'         # confusion-matrix colour map


def set_seed(seed=SEED):
    """Fix all random seeds so runs are repeatable."""
    torch.manual_seed(seed)
    np.random.seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)


set_seed()


# ---------------------------- PATH RESOLUTION --------------------------------
IMG_EXTS = ('.png', '.jpg', '.jpeg', '.tiff', '.tif')


def find_tissue_root(configured_path, base=KAGGLE_INPUT):
    """
    Return the folder that directly contains the class sub-folders
    (01_TUMOR, 02_STROMA, ...). Uses the configured path if valid,
    otherwise searches everything under /kaggle/input.
    """
    if os.path.isdir(os.path.join(configured_path, '01_TUMOR')):
        return configured_path
    for root, dirs, _ in os.walk(base):
        if '01_TUMOR' in dirs:
            print(f"Auto-detected tissue dataset at: {root}")
            return root
    raise FileNotFoundError(
        f"Could not find '01_TUMOR' under {base}. Check that the dataset is attached.")


def find_unknown_root(configured_path=None, base=KAGGLE_INPUT):
    """
    Return a folder of random images for the UNKNOWN class, or None if absent.
    Looks for a directory with 'unsplash' in its path that holds images directly.
    """
    if configured_path and os.path.isdir(configured_path):
        return configured_path
    for root, _, files in os.walk(base):
        if 'unsplash' in root.lower() and any(f.lower().endswith(IMG_EXTS) for f in files):
            print(f"Auto-detected UNKNOWN images at: {root}")
            return root
    print("WARNING: no Unsplash folder found - training WITHOUT the UNKNOWN class images.")
    return None


# ------------------------------ 3. DATASET -----------------------------------
class ColorectalCancerDataset(Dataset):
    """
    Loads the 8 colorectal tissue classes (one folder per class) and,
    optionally, a flat folder of random photos labelled as 'UNKNOWN'.
    """

    IMG_EXTENSIONS = ('.png', '.jpg', '.jpeg', '.tiff', '.tif')

    def __init__(self, root_dir, transform=None, unknown_dir=None, max_unknown=None):
        self.root_dir = root_dir
        self.unknown_dir = unknown_dir
        self.transform = transform

        # 8 tissue classes + 1 out-of-distribution class
        self.classes = [
            '01_TUMOR', '02_STROMA', '03_COMPLEX', '04_LYMPHO',
            '05_DEBRIS', '06_MUCOSA', '07_ADIPOSE', '08_EMPTY',
            'UNKNOWN'
        ]
        self.class_to_idx = {cls: idx for idx, cls in enumerate(self.classes)}

        self.samples = []                 # list of (image_path, label)
        self.class_counts = Counter()

        self._load_tissue_classes()
        self._load_unknown_class(max_unknown)

        if len(self.samples) == 0:
            raise RuntimeError(f"No images found under '{root_dir}'. Check the dataset path.")

        print("Class counts:", dict(self.class_counts))
        print("Total samples:", len(self.samples))

    def _load_tissue_classes(self):
        """Read images from each of the 8 class sub-folders."""
        for class_name in self.classes:
            if class_name == 'UNKNOWN':
                continue
            class_path = os.path.join(self.root_dir, class_name)
            if not os.path.exists(class_path):
                continue
            items = [(os.path.join(class_path, f), self.class_to_idx[class_name])
                     for f in os.listdir(class_path)
                     if f.lower().endswith(self.IMG_EXTENSIONS)]
            self.samples.extend(items)
            self.class_counts[class_name] = len(items)

    def _load_unknown_class(self, max_unknown):
        """Read the flat folder of random images and label them UNKNOWN."""
        if self.unknown_dir is None or not os.path.exists(self.unknown_dir):
            return
        items = [(os.path.join(self.unknown_dir, f), self.class_to_idx['UNKNOWN'])
                 for f in os.listdir(self.unknown_dir)
                 if f.lower().endswith(self.IMG_EXTENSIONS)]
        if max_unknown is not None:
            items = items[:max_unknown]
        self.samples.extend(items)
        self.class_counts['UNKNOWN'] = len(items)

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_path, label = self.samples[idx]
        try:
            image = Image.open(img_path).convert('RGB')
        except Exception as e:
            # Corrupted file -> fall back to a black placeholder image
            print(f"Error loading image {img_path}: {e}")
            image = Image.new('RGB', (224, 224), color='black')
        if self.transform:
            image = self.transform(image)
        return image, label


class TransformedSubset(Dataset):
    """
    Wraps a subset (train/val/test) and applies its OWN transform.
    Needed because random_split subsets share one underlying dataset, so
    setting `.dataset.transform` would overwrite the transform for all splits.
    """

    def __init__(self, subset, transform):
        self.subset = subset
        self.transform = transform

    def __len__(self):
        return len(self.subset)

    def __getitem__(self, idx):
        image, label = self.subset[idx]      # raw PIL image (no transform yet)
        return self.transform(image), label


# ------------------------ 4. TRANSFORMS & DATA LOADERS -----------------------
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]


def get_transforms():
    """Augmented transform for training; plain resize + normalise for eval."""
    train_transform = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomVerticalFlip(p=0.5),
        transforms.RandomRotation(degrees=15),
        transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.1),
        transforms.RandomAffine(degrees=0, translate=(0.1, 0.1), scale=(0.9, 1.1)),
        transforms.RandomApply([transforms.GaussianBlur(kernel_size=3)], p=0.2),
        transforms.ToTensor(),
        transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    ])
    eval_transform = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    ])
    return train_transform, eval_transform


def create_data_loaders(dataset_dir, batch_size=32, train_ratio=0.7, val_ratio=0.15,
                        unknown_dir=None, max_unknown=None):
    """Build train / validation / test loaders using a 70-15-15 split."""
    print("Creating data loaders with 70-15-15 split...")
    train_transform, eval_transform = get_transforms()

    # Load the dataset once without transforms; transforms are added per split
    full_dataset = ColorectalCancerDataset(
        dataset_dir, transform=None, unknown_dir=unknown_dir, max_unknown=max_unknown
    )

    # Compute split sizes (test gets the remainder to avoid rounding loss)
    total = len(full_dataset)
    train_size = int(train_ratio * total)
    val_size = int(val_ratio * total)
    test_size = total - train_size - val_size

    generator = torch.Generator().manual_seed(SEED)
    train_sub, val_sub, test_sub = random_split(
        full_dataset, [train_size, val_size, test_size], generator=generator
    )

    # Attach the correct transform to each split independently
    train_ds = TransformedSubset(train_sub, train_transform)
    val_ds = TransformedSubset(val_sub, eval_transform)
    test_ds = TransformedSubset(test_sub, eval_transform)

    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True,
                              num_workers=0, pin_memory=True, drop_last=True)
    val_loader = DataLoader(val_ds, batch_size=batch_size, shuffle=False,
                            num_workers=0, pin_memory=True)
    test_loader = DataLoader(test_ds, batch_size=batch_size, shuffle=False,
                             num_workers=0, pin_memory=True)

    print(f"Train size: {len(train_ds)}, Val size: {len(val_ds)}, Test size: {len(test_ds)}")
    return train_loader, val_loader, test_loader, full_dataset.classes


# ------------------------------- 5. MODEL ------------------------------------
class DenseNetClassifier(nn.Module):
    """DenseNet121 (ImageNet pretrained) with a custom multi-layer head."""

    def __init__(self, num_classes=9, dropout_rate=0.5):
        super().__init__()

        # Load ImageNet-pretrained backbone
        try:
            self.backbone = models.densenet121(weights="IMAGENET1K_V1")
            print("DenseNet121 with IMAGENET1K_V1 weights loaded successfully")
        except Exception as e:
            print("Error loading DenseNet121 pretrained weights:", e)
            raise

        # Freeze everything except the last 30 parameter tensors (stable fine-tuning)
        for param in list(self.backbone.parameters())[:-30]:
            param.requires_grad = False

        # Replace the original classifier with a deeper, regularised head
        in_features = self.backbone.classifier.in_features
        self.backbone.classifier = nn.Sequential(
            nn.Dropout(p=dropout_rate),
            nn.Linear(in_features, 512),
            nn.ReLU(inplace=True),
            nn.BatchNorm1d(512),
            nn.Dropout(p=dropout_rate * 0.5),
            nn.Linear(512, 256),
            nn.ReLU(inplace=True),
            nn.BatchNorm1d(256),
            nn.Dropout(p=dropout_rate * 0.3),
            nn.Linear(256, num_classes),
        )

    def forward(self, x):
        return self.backbone(x)


# ------------------------- 6. TRAINING UTILITIES -----------------------------
class EarlyStopping:
    """Stop training when validation loss stops improving."""

    def __init__(self, patience=10, min_delta=0.001, restore_best_weights=True):
        self.patience = patience
        self.min_delta = min_delta
        self.restore_best_weights = restore_best_weights
        self.best_loss = None
        self.counter = 0
        self.best_weights = None

    def __call__(self, val_loss, model):
        if self.best_loss is None or val_loss < self.best_loss - self.min_delta:
            # Improvement -> remember this state and reset the patience counter
            self.best_loss = val_loss
            self.counter = 0
            self.save_checkpoint(model)
        else:
            self.counter += 1

        if self.counter >= self.patience:
            if self.restore_best_weights and self.best_weights is not None:
                model.load_state_dict(self.best_weights)
            return True
        return False

    def save_checkpoint(self, model):
        self.best_weights = {k: v.clone() for k, v in model.state_dict().items()}


def run_epoch(model, loader, criterion, device, desc, optimizer=None):
    """
    Run one pass over `loader`.
    If an optimizer is given -> training mode, otherwise evaluation mode.
    Returns (average_loss, accuracy_percent).
    """
    training = optimizer is not None
    model.train() if training else model.eval()

    total_loss, correct, total = 0.0, 0, 0
    pbar = tqdm(loader, desc=desc)

    with torch.set_grad_enabled(training):
        for data, target in pbar:
            data = data.to(next(model.parameters()).device, non_blocking=True)
            target = target.to(data.device, non_blocking=True)

            if training:
                optimizer.zero_grad()

            output = model(data)
            loss = criterion(output, target)

            if training:
                loss.backward()
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)  # avoid exploding gradients
                optimizer.step()

            total_loss += loss.item()
            correct += (output.argmax(1) == target).sum().item()
            total += target.size(0)
            pbar.set_postfix({'Loss': f'{loss.item():.4f}',
                              'Acc': f'{100. * correct / total:.2f}%'})

    return total_loss / len(loader), 100. * correct / total


def train_model(model, train_loader, val_loader, device, num_epochs=40):
    """Full training loop with label smoothing, cosine LR and early stopping."""
    criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
    optimizer = optim.AdamW(model.parameters(), lr=0.001, weight_decay=0.01)
    scheduler = CosineAnnealingLR(optimizer, T_max=num_epochs, eta_min=1e-6)
    early_stopping = EarlyStopping(patience=10, min_delta=0.001)

    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}
    best_val_acc = 0.0

    for epoch in range(num_epochs):
        tag = f'Epoch {epoch + 1}/{num_epochs}'

        # --- training pass, then validation pass ---
        train_loss, train_acc = run_epoch(model, train_loader, criterion, device,
                                          f'{tag} [Train]', optimizer)
        val_loss, val_acc = run_epoch(model, val_loader, criterion, device,
                                      f'{tag} [Val]')

        # --- record metrics ---
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)

        print(f'\n{tag} results:')
        print(f'   Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}%')
        print(f'   Val   Loss: {val_loss:.4f} | Val   Acc: {val_acc:.2f}%')

        # --- checkpoint the best model by validation accuracy ---
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            torch.save(model.state_dict(), 'best_colorectal_model.pth')
            print(f'   New best model saved! Val Acc: {best_val_acc:.2f}%')
        print('-' * 70)

        scheduler.step()

        # --- stop early if validation loss has plateaued ---
        if early_stopping(val_loss, model):
            print(f'Early stopping triggered after epoch {epoch + 1}')
            break

        # --- periodic memory cleanup ---
        if epoch % 10 == 0:
            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

    return history, best_val_acc


# --------------------------- 7. EVALUATION -----------------------------------
def evaluate_model(model, test_loader, device):
    """Return predictions, ground-truth labels and class probabilities."""
    model.eval()
    all_preds, all_targets, all_probs = [], [], []

    with torch.no_grad():
        for data, target in tqdm(test_loader, desc='Evaluating'):
            data = data.to(device, non_blocking=True)
            output = model(data)
            probs = torch.softmax(output, dim=1)

            all_preds.extend(output.argmax(1).cpu().numpy())
            all_targets.extend(target.numpy())
            all_probs.extend(probs.cpu().numpy())

    return np.array(all_preds), np.array(all_targets), np.array(all_probs)


# ----------------------------- 8. PLOTTING -----------------------------------
def plot_training_history(history):
    """Side-by-side accuracy and loss curves for train vs validation."""
    plt.style.use('seaborn-v0_8-whitegrid')
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    epochs = range(1, len(history['train_acc']) + 1)

    # Accuracy panel
    ax1.plot(epochs, history['train_acc'], color=PALETTE['train'], lw=2.5,
             marker='o', ms=4, label='Training Accuracy')
    ax1.plot(epochs, history['val_acc'], color=PALETTE['val'], lw=2.5,
             marker='s', ms=4, label='Validation Accuracy')
    ax1.set_title('Model Accuracy', fontsize=14, fontweight='bold')
    ax1.set_xlabel('Epoch', fontsize=12)
    ax1.set_ylabel('Accuracy (%)', fontsize=12)
    ax1.set_ylim([0, 100])
    ax1.legend(fontsize=10)

    # Loss panel
    ax2.plot(epochs, history['train_loss'], color=PALETTE['train'], lw=2.5,
             marker='o', ms=4, label='Training Loss')
    ax2.plot(epochs, history['val_loss'], color=PALETTE['val'], lw=2.5,
             marker='s', ms=4, label='Validation Loss')
    ax2.set_title('Model Loss', fontsize=14, fontweight='bold')
    ax2.set_xlabel('Epoch', fontsize=12)
    ax2.set_ylabel('Loss', fontsize=12)
    ax2.legend(fontsize=10)

    plt.tight_layout()
    plt.show()


def plot_confusion_matrix(y_true, y_pred, class_names):
    """Heatmap of true vs predicted labels."""
    cm = confusion_matrix(y_true, y_pred)

    plt.figure(figsize=(12, 10))
    sns.heatmap(cm, annot=True, fmt='d', cmap=CM_CMAP,
                linewidths=0.5, linecolor='white',
                xticklabels=class_names, yticklabels=class_names,
                cbar_kws={'label': 'Number of Samples'})
    plt.title('Confusion Matrix', fontsize=16, fontweight='bold', pad=20)
    plt.xlabel('Predicted Label', fontsize=12)
    plt.ylabel('True Label', fontsize=12)
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
    return cm


def plot_roc_curves(y_true, y_probs, class_names):
    """One-vs-rest ROC curve per class; returns the mean AUC."""
    n_classes = len(class_names)
    y_true_bin = label_binarize(y_true, classes=range(n_classes))

    plt.figure(figsize=(14, 10))
    mean_auc = 0.0

    for i, class_name in enumerate(class_names):
        fpr, tpr, _ = roc_curve(y_true_bin[:, i], y_probs[:, i])
        class_auc = auc(fpr, tpr)
        mean_auc += class_auc
        plt.plot(fpr, tpr, color=ROC_COLORS[i % len(ROC_COLORS)], lw=2.5,
                 label=f'{class_name} (AUC = {class_auc:.3f})')

    mean_auc /= n_classes

    # Diagonal = performance of random guessing
    plt.plot([0, 1], [0, 1], linestyle='--', lw=2, color=PALETTE['chance'],
             label='Random Classifier (AUC = 0.500)')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate', fontsize=12)
    plt.ylabel('True Positive Rate', fontsize=12)
    plt.title(f'ROC Curves - Multi-class Classification\nMean AUC = {mean_auc:.3f}',
              fontsize=14, fontweight='bold')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=10)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
    return mean_auc


# ------------------------------- 9. MAIN -------------------------------------
def describe_kappa(kappa):
    """Translate Cohen's Kappa into a plain-English agreement level."""
    if kappa > 0.8:
        return "Excellent agreement"
    if kappa > 0.6:
        return "Substantial agreement"
    if kappa > 0.4:
        return "Moderate agreement"
    return "Poor agreement"


def main():
    print("Starting Colorectal Cancer Classification with DenseNet121 + UNKNOWN class")
    print("=" * 60)

    # ---- Device ----
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"Using device: {device}")

    # ---- Data ----
    print("\n" + "=" * 60)
    tissue_dir = find_tissue_root(DATASET_DIR)
    unknown_dir = find_unknown_root(UNSPLASH_DIR)
    train_loader, val_loader, test_loader, class_names = create_data_loaders(
        tissue_dir, BATCH_SIZE,
        train_ratio=0.7, val_ratio=0.15,
        unknown_dir=unknown_dir, max_unknown=MAX_UNKNOWN,
    )
    print(f"\nClasses: {class_names}")

    # ---- Model ----
    print("\nBuilding DenseNet121 model...")
    model = DenseNetClassifier(num_classes=len(class_names), dropout_rate=0.5).to(device)
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"Total params: {total_params:,} | Trainable params: {trainable_params:,}")

    # ---- Training ----
    print("\n" + "=" * 60)
    print("Starting training...")
    history, best_val_acc = train_model(model, train_loader, val_loader, device, NUM_EPOCHS)

    # Reload the best checkpoint before testing
    model.load_state_dict(torch.load('best_colorectal_model.pth', map_location=device))
    print(f"\nBest validation accuracy: {best_val_acc:.2f}%")

    print("\nPlotting training history...")
    plot_training_history(history)

    # ---- Test-set evaluation ----
    print("\n" + "=" * 60)
    print("Evaluating on test set...")
    y_pred, y_true, y_probs = evaluate_model(model, test_loader, device)

    test_accuracy = (y_pred == y_true).mean() * 100
    print(f"\nFinal Test Accuracy: {test_accuracy:.2f}%")

    print("\nClassification Report:")
    print("=" * 80)
    print(classification_report(y_true, y_pred, target_names=class_names, digits=4))

    kappa = cohen_kappa_score(y_true, y_pred)
    print(f"\nCohen's Kappa Score: {kappa:.4f} -> {describe_kappa(kappa)}")

    # ---- Visualisations ----
    print("\nGenerating visualizations...")
    plot_confusion_matrix(y_true, y_pred, class_names)
    mean_auc = plot_roc_curves(y_true, y_probs, class_names)

    # ---- Summary ----
    print("\n" + "=" * 60)
    print("FINAL RESULTS SUMMARY")
    print("=" * 60)
    print(f"Test Accuracy:     {test_accuracy:.2f}%")
    print(f"Mean AUC Score:    {mean_auc:.3f}")
    print(f"Cohen's Kappa:     {kappa:.4f}")
    print(f"Best Val Accuracy: {best_val_acc:.2f}%")

    # ---- Save final model together with its metadata ----
    torch.save({
        'model_state_dict': model.state_dict(),
        'class_names': class_names,
        'test_accuracy': test_accuracy,
        'cohen_kappa': kappa,
        'mean_auc': mean_auc,
        'history': history,
    }, 'final_colorectal_model.pth')

    print("\nModels saved successfully!")
    print("   - best_colorectal_model.pth (best weights)")
    print("   - final_colorectal_model.pth (with metadata)")


if __name__ == "__main__":
    main()
```
